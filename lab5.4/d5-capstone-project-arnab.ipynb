### Part 3: Build Models to train the dataset using different Machine Learning algorithms, and document your architecture choice

**Task 3:**

```python
# x_test,x_train,y_test,y_train=train_test_split(x,y,test_size=0.1,random_state=0),58

x_train1, x_test1, y_train1, y_test1 = train_test_split(x, y, test_size=0.1, random_state=72)
model = LogisticRegression(random_state=58)
model.fit(x_train1, y_train1)
y_pred1 = model.predict(x_test1)

accuracy1 = accuracy_score(y_test1, y_pred1)
classification_rep1=classification_report(y_test1, y_pred1)


print(" Accuracy in LOG_REG:", accuracy1)
print(f'Classification Report:\n{classification_rep1}')
```

```python
# # Decesion tree classification
x_train2, x_test2, y_train2, y_test2 = train_test_split(x, y, test_size=0.1, random_state=72)
model2 = DecisionTreeClassifier(random_state=72)
model2.fit(x_train2, y_train2)
y_pred2 = model2.predict(x_test2)

accuracy2 = accuracy_score(y_test2, y_pred2)
classification_rep2=classification_report(y_test2, y_pred2)


print("Accuracy in DEC_TREE:", accuracy2)
print(f'Classification Report:\n{classification_rep2}')
```

```python

x_train3,x_test3,y_train3,y_test3=train_test_split(x,y, test_size=0.1,random_state=72)
model3=SVC(random_state=72)
model3.fit(x_train3,y_train3)
y_pred3=model3.predict(x_test3)

accuracy3=accuracy_score(y_test3,y_pred3)
classification_rep3=classification_report(y_test3,y_pred3)

print("Accuracy in SVM:", accuracy3)
print(f'Classification Report:\n{classification_rep3}')
```

```python
x_train4,x_test4,y_train4,y_test4=train_test_split(x,y, test_size=0.1,random_state=72)
model4=RandomForestClassifier(random_state=72)
model4.fit(x_train4,y_train4)
y_pred4=model4.predict(x_test4)

accuracy4=accuracy_score(y_test4,y_pred4)
classification_rep4=classification_report(y_test4,y_pred4)

print("Accuracy in RF:", accuracy4)
print(f'Classification Report:\n{classification_rep4}')
```

### Part 4: Evaluating Model & Business Justification

**Task 4:**
- Analyze model performance metrics to identify the most effective algorithm.
- Translate that performance into a business justification: estimate the cost of readmissions this model could help avoid, and weigh the trade-off between false positives (unnecessary interventions) and false negatives (missed at-risk patients).

```python
# Illustrative assumptions for the exercise; validate these with hospital finance
# and care-management teams before using them in a real business case.
avg_cost_per_readmission = 15_000
cost_per_intervention = 300
assumed_prevention_rate = 0.30

true_positives = 14
false_positives = 17
false_negatives = 10

interventions_triggered = true_positives + false_positives
intervention_cost = interventions_triggered * cost_per_intervention
readmissions_prevented = true_positives * assumed_prevention_rate
estimated_savings = readmissions_prevented * avg_cost_per_readmission
net_benefit = estimated_savings - intervention_cost

print(f"Interventions triggered: {interventions_triggered}")
print(f"Intervention cost: ${intervention_cost:,.0f}")
print(f"Estimated readmissions prevented: {readmissions_prevented:.1f}")
print(f"Estimated savings: ${estimated_savings:,.0f}")
print(f"Estimated net benefit: ${net_benefit:,.0f}")
print(f"Missed readmitted patients: {false_negatives}")
```

## Part 5: Implementation Roadmap

| Phase | Timing | Activities and deliverables | Decision gate |
|---|---:|---|---|
| **1. Validate data and workflow** | Weeks 0–2 | Confirm the readmission outcome definition and prediction point are appropriate for the intended use. Review data quality, feature availability at discharge, privacy/access requirements, and how care managers would receive and act on a risk flag. Agree on pilot measures and baseline. | Proceed only if the label, data, and proposed workflow are reliable and clinically appropriate. |
| **2. Strengthen model validation** | Weeks 2–4 | Re-run evaluation with a reproducible split and compare against a simple baseline. Assess recall, precision, calibration, and performance at the alert volume the team can manage. Check results across relevant patient subgroups where sample sizes allow. | Do not advance to live use if performance is unstable, materially inequitable, or not useful at feasible alert volumes. |
| **3. Silent-mode test** | Weeks 4–6 | Generate scores prospectively without showing them to staff or changing care. Check that scoring happens before discharge, inputs are available and accurate, and score distributions and performance are consistent with expectations. | Resolve workflow, data, and reliability issues before exposing scores to users. |
| **4. Human-reviewed pilot** | Weeks 6–10 | Show the risk flag to a small care-management team. Keep clinical judgment in control; document whether alerts are reviewed, what follow-up occurs, staff workload, and patient outcomes. Monitor for missed cases and excessive alerts. | Continue only if the workflow is usable and safety, workload, and outcome measures are acceptable. |
| **5. Evaluate and decide on scale-up** | After the pilot | Compare pilot results with the agreed baseline and, if feasible, an appropriate comparison group. Review readmissions, follow-up completion, intervention uptake, staff burden, patient experience, equity, and net costs. Retrain or revise the model only with validated data and documented review. | Scale, revise, or stop based on measured results—not the illustrative savings estimate alone. |

**Key risks and mitigations:** The training sample is small and reported model performance is modest, so treat the model as an unvalidated decision-support signal. Confirm that all features are available before discharge to avoid data leakage. Monitor false positives so alerts do not overwhelm staff, and false negatives so the model does not create misplaced confidence. Review performance and access across patient groups; if subgroup results are too sparse to assess, keep that limitation explicit and do not assume equitable performance.
